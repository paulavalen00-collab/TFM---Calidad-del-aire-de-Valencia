## ZONAS VERDES CARGA

Leo csv generado desde QGIS y lo transformo en parquet

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder \
    .appName("TFM_Zonas_Verdes") \
    .getOrCreate()
```

```python
from pyspark.sql import functions as F

verde = (
    spark.read.option("header", True)
    .csv("gs://tfm_unir1/verdes/verde_distrito.csv")
    .select(
        F.col("coddistrit").cast("int").alias("id_distrito"),
        F.round(F.col("sum").cast("double"), 0).alias("m2_verde"),
        F.col("count").cast("int").alias("n_espacios"),
    )
    .filter(F.col("id_distrito").isNotNull())
)

verde.orderBy("id_distrito").show()   # 19 filas

verde.write.mode("overwrite").parquet("gs://tfm_unir1/processed/verde_distrito/")
print("Guardado")
```
