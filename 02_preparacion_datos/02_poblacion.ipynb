## POBLACION POR DISTRITOS

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder \
    .appName("TFM_Genericos_Parquets") \
    .getOrCreate()
```

```python
from pyspark.sql import functions as F

poblacion = (
    spark.read.option("header", True)
    .csv("gs://tfm_unir1/poblacion/poblacion_distrito.csv")
    .select(
        F.col("id_distrito").cast("int").alias("id_distrito"),
        F.col("distrito").alias("distrito"),              
        F.col("anio").cast("int").alias("anio"),
        F.col("poblacion").cast("int").alias("poblacion"),
    )
)

# Comprobación: 19 distritos por año
poblacion.groupBy("anio").agg(
    F.count("*").alias("distritos"),
    F.sum("poblacion").alias("total")
).orderBy("anio").show()
# 2023 -> 19 | 809501
# 2024 -> 19 | 830606
# 2025 -> 19 | 844424

poblacion.write.mode("overwrite").parquet("gs://tfm_unir1/processed/poblacion_distrito/")
print("Guardado")
```
